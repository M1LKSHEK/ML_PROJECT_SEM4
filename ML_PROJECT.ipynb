```python
import pandas as pd
import numpy as np
import os
import glob

found_paths = glob.glob('/content/**/results.csv', recursive=True)

if found_paths:
    DATA_PATH = os.path.dirname(found_paths[0])
    print(f"🎉 Success! Found your F1 files inside: {DATA_PATH}\n")
else:
    DATA_PATH = '/content'
    print("⚠️ Could not find 'results.csv' automatically.")
    print("Please make sure your files are uploaded. Current main folder contents:", os.listdir('/content'))


try:
    results = pd.read_csv(os.path.join(DATA_PATH, 'results.csv'))
    races = pd.read_csv(os.path.join(DATA_PATH, 'races.csv'))
    drivers = pd.read_csv(os.path.join(DATA_PATH, 'drivers.csv'))
    constructors = pd.read_csv(os.path.join(DATA_PATH, 'constructors.csv'))

    circuits = pd.read_csv(os.path.join(DATA_PATH, 'circuits.csv'), encoding='latin-1')

    df = pd.merge(results, races, on='race_id', how='inner')
    df = pd.merge(df, drivers, on='driver_id', how='inner', suffixes=('', '_driver'))
    df = pd.merge(df, constructors, on='constructor_id', how='inner', suffixes=('', '_constructor'))
    df_master = pd.merge(df, circuits, on='circuit_id', how='inner', suffixes=('', '_circuit'))

    print("✅ Master Dataset Successfully Created!")
    print(f"Total Rows & Columns: {df_master.shape}\n")

    print(df_master[['season', 'race_name', 'familyName', 'name', 'grid', 'position_order']].head())

except Exception as e:
    print(f"❌ An error occurred: {e}")
```
